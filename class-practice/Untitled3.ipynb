```python
python = int(input("Enter python marks"))
numpy = int(input("Enter numpy marks"))
pandas = int(input("Enter pasdas marks"))
stats = int(input("Enter stats marks"))
mL = int(input("Enter mL marks"))
total = python + numpy + pandas + stats + mL
percentage = total / 500 * 100
result = None
grade = None
if python >= 50:
  if numpy >= 50:
    if pandas >= 50:
      if stats >= 50:
        if mL >= 50:
          result = "Pass"
        else:
          result = "Fail"
      else:
        result = "Fail"
    else:
      result = "Fail"
  else:
    result = "Fail"
else:
  result = "Fail"
if result == "Pass":
  if percentage >= 90:
    grade = "A+"
  elif percentage >= 80:
    grade = "A"
  elif percentage >= 70:
    grade = "B"
  elif percentage >= 60:
    grade = "C"
  elif percentage >= 50:
    grade = "D"
  else:
    grade = "F"
print("Total", total)
print("Percentage", percentage)
print("Result", result)
print("Grade", grade)
```
