```python
store = {
    'sugar': {'price':200, 'quantity': 1000},
    'Rice': {'price':400, 'quantity': 500},
    'meat': {'price':600, 'quantity': 300},
    'salt': {'price':2000, 'quantity': 100},
    'Flour': {'price':700, 'quantity': 5000}
}
```

```python
item = input('Do you have ').lower()
for product_item in store:
  if product_item == item:
    print(f"{item} available hy.")
    print(f"{store[product_item]['price']} rupey kilo he.")
    kgs = int(input("How many KGs"))
    if kgs<=store[product_item]['quantity']:
      bill = store[product_item]['price'] * kgs
      store[product_item]['quantity'] -= kgs
      print(f"please pay amount {bill}")
    else:
      print(f"{kgs} kgs {product_item} not available")

    break
else:
  print('Not Available')
```
